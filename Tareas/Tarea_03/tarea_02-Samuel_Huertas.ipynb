# 1.

Demuestre que el problema de valor inicial: 

$$

\begin{cases}
u_t(t,x) = b\,u_{xx}(t,x), & 0 < x < \ell,\quad t > 0, \\[6pt]
u(0,x) = u_0(x), & 0 \le x \le \ell,
\end{cases}

$$

donde $b \in  \mathbb{R}^+$ y $u_0$ funciónn conocida, esta “bien propuesto”

El PVI está bien puesto si este cumple:  

$ \forall T > 0  \exist C_T > 0 $ talque:

$$
\int_{-\infty}^{\infty} |u(t,x)|^2 dx \leq C_T \int_{-\infty}^{\infty} |u(0,x)|^2 dx, \forall 0 \leq t \leq T
$$


Utilizando la transformada de Fourier en la parte espacial, vemos que:

$$
\hat{u}_t(t, \omega) = -b \omega^2 \hat{u}(t, \omega)
$$

esto sale de las propiedades de la transformada de Fourier, la cual cumple:

* $F[u_t(t,x)] = \hat{u}_t(t,\omega)$

* $F[u_{xx}(t,x)] = (i \omega)^2\hat{u}(t,\omega) = -\omega^2\hat{u}(t,\omega)$ , con $F[\frac{d^n}{dx^n} u(t,x)] = (i \omega)^n \hat{u}(t,w)$

Utilizando los valores iniciales la solución a la ecuación queda de la forma:

$$
\hat{u}(t, \omega) = e^{ -b \omega^2 t} \hat{u}_0(\omega)
$$

Utilizando la transformada inversa de Fourier: 

$$
u(t, x) = \frac{1}{\sqrt{2 \pi}} \int_{-\infty}^{\infty} e^{i \omega x} e^{ -b \omega^2 t} \hat{u}_0(\omega) d\omega
$$

Utilizando la transformada inversa de Fourier, obtenesmos:

$$\hat{u}_0(\omega) = \frac{1}{\sqrt{2\pi}} \int_{-\infty}^{\infty} e^{-i\omega y} u_0(y) \, dy$$

Lo cual nos queda: 

$$
u(t, x) = \frac{1}{\sqrt{2 \pi}} \int_{-\infty}^{\infty} e^{i \omega x} e^{ -b \omega^2 t} \left[\frac{1}{\sqrt{2\pi}} \int_{-\infty}^{\infty} e^{-i\omega y} u_0(y) \, dy \right] d\omega
$$

Organizando los terminos nos queda:

$$u(t,x) = \frac{1}{\sqrt{4\pi}} \int_{-\infty}^{\infty} \left( \frac{1}{\sqrt{\pi}} \int_{-\infty}^{\infty} e^{i\omega(x-y)} e^{-b\omega^2t} d\omega \right) u_0(y) \, dy$$

Mirando la integral entre los parentesis, tenemos que: 

$$
 \frac{1}{\sqrt{\pi}} \int_{-\infty}^{\infty} e^{i\omega(x-y)-bt\omega^2} d\omega
$$




Completando cuadrados en la exponencial: 

<center>

$i\omega(x-y)-bt\omega^2 = - bt \left[ \omega^2 - \frac{i (x-y)}{bt}\omega \right]  $ 

</center>

Llamando $\beta = \frac{i (x-y)}{bt}$:

$ - bt \left[ \omega^2 - \frac{i (x-y)}{bt}\omega \right] =  - bt \left[ \omega^2 - \beta \omega \right]  =  - bt \left[ \omega^2 - \beta \omega  + \frac{\beta^2}{4} - \frac{\beta^2}{4} \right] = - bt \left[ \left(\omega - \frac{\beta}{2} \right)^2 - \frac{\beta^2}{4} \right] = - bt \left(\omega - \frac{\beta}{2} \right)^2 + bt \frac{\beta^2}{4}$

Sustituyendo en la integral se obtiene: 

$$
 \frac{1}{\sqrt{\pi}} \int_{-\infty}^{\infty} e^{- bt \left(\omega - \frac{\beta}{2} \right)^2 + bt \frac{\beta^2}{4}} d\omega
$$

Reemplazando $\beta$:

$$
 \frac{1}{\sqrt{\pi}} \int_{-\infty}^{\infty} e^{- bt \left(\omega - \frac{i (x-y)}{2bt} \right)^2 + \frac{i^2 (x-y)^2}{4bt} } d\omega
$$


Reorganizando nos queda: 

$$
 \frac{e^{\frac{- (x-y)^2}{4bt}}}{\sqrt{\pi}} \int_{-\infty}^{\infty} e^{- bt \left(\omega - \frac{i (x-y)}{2bt} \right)^2} d\omega
$$

La integral se conoce como una integral gaussiana la cual cumple con: 

$$
\int_{-\infty}^{\infty} e^{- a x^2} dx = \sqrt{\frac{\pi}{a}}
$$


Realizando la sustitución de $q = \omega - \frac{i (x-y)}{2bt}$; $dq = d\omega$ y vemos que los limites sde integración no van a cambiar, por lo que la integral nos queda: 

$$
 \frac{e^{\frac{- (x-y)^2}{4bt}}}{\sqrt{\pi}} \int_{-\infty}^{\infty} e^{- bt \left(q \right)^2} dq = \frac{e^{\frac{- (x-y)^2}{4bt}}}{\sqrt{\pi}} \sqrt{\frac{\pi}{bt}} = \frac{e^{\frac{- (x-y)^2}{4bt}}}{\sqrt{bt}} 
$$

Dando como resultado: 

$$u(t, x) = \frac{1}{\sqrt{4\pi bt}} \int_{-\infty}^{\infty} e^{-(x-y)^2/4bt} u_0(y) \, dy$$

La anterior expresión nos muestra que $u(t,x)$ se puedo tomar como medida ponderada de la función $u_0$. También se puede notar que para tiempos $t$ muy pequeños la función de pesos que multiplica a $u_0(y)$ tiene un pico en $y=x$, mientras que para tiempos grandes la función de pesos es mucho más amplia. 

Al mirar $|u(t,x)|^2$ y aplicar la desigualdad de Cauchy–Schwarz, la cual es:

$$
\left| \int_{\mathbb{R}^n} f(x)\, \overline{g(x)} \, dx \right|^2
\leq 
\int_{\mathbb{R}^n} |f(x)|^2 \, dx 
\int_{\mathbb{R}^n} |g(x)|^2 \, dx
$$


Vemos que:

$$|u(t, x)|^2 =  \frac{1}{\sqrt{4\pi bt}} \left|\int_{-\infty}^{\infty} e^{-(x-y)^2/4bt} u_0(y) \, dy \right|^2 \leq \frac{1}{\sqrt{4\pi bt}} \int_{-\infty}^{\infty} |e^{-(x-y)^2/4bt}|^2 dy  \int_{-\infty}^{\infty} |u_0(y)|^2 \, dy$$

Tenemos que para valores positivos de $t$ la integral: 

$$
\int_{-\infty}^{\infty} |e^{-(x-y)^2/4bt}|^2 dy
$$

Pertenece a $L^2(R)$, con lo que se puede decir que 

$$

|u(t, x)|^2  \leq C_T  \int_{-\infty}^{\infty} |u_0(y)|^2 \, dy

$$

Por lo que al realizar la integral con respecto a $x$, siendo $y$ una varaible muda, tenemos que:

$$

\int_{-\infty}^{\infty}|u(t, x)|^2 dx \leq C_T  \int_{-\infty}^{\infty} \int_{-\infty}^{\infty} |u_0(y)|^2 \, dy \, dx

$$

Por lo tanto podemos decir que el problema está "bien puesto".

\frac{\varphi_{m}^{n+1} - (\varphi)_{m}^{n}}{k} =  (\varphi_{t})_{m}^{n} + k/2 (\varphi_{tt})_{m}^{n} + O(k^2)
$

</center>

* La parte espacial $ \frac{\varphi_{m+1}^{n} - 2\varphi_{m}^{n} + \varphi_{m-1}^{n}}{2h^2}$:

<center>

$
\varphi_{m \pm 1}^{n} = (\varphi)_{m}^{n} \pm h (\varphi_{x})_{m}^{n} + h^2/2 (\varphi_{xx})_{m}^{n} + O(h^3)

$

$
\varphi_{m+1}^{n} - 2 (\varphi)_{m}^{n} + \varphi_{m-1}^{n} = 2h^2/2 (\varphi_{xx})_{m}^{n} + O(h^4)
$

$
\frac{\varphi_{m+1}^{n} - 2 (\varphi)_{m}^{n} + \varphi_{m-1}^{n}}{2h^2} =  1/2 (\varphi_{xx})_{m}^{n} + O(h^2)
$

</center>

* Mirando el termino combinado $\frac{\varphi_{m+1}^{n+1} - 2\varphi_{m}^{n+1} + \varphi_{m-1}^{n+1} }{2h^2} $: 

<center>

$
\frac{\varphi_{m+1}^{n+1} - 2\varphi_{m}^{n+1} + \varphi_{m-1}^{n+1}}{2h^2} =  1/2 (\varphi_{xx})_{m}^{n+1} + O(h^2)
$

</center>

Para el termino $ (\varphi_{xx})_{m}^{n+1}$:

<center>

$
(\varphi_{xx})_{m}^{n+1} = (\varphi_{xx})_{m}^{n} + k (\varphi_{xxt})_{m}^{n} + O(k^2)
$

</center>

Con lo que nos queda: 

<center>

$
\frac{\varphi_{m+1}^{n+1} - 2\varphi_{m}^{n+1} + \varphi_{m-1}^{n+1}}{2h^2} =  1/2  (\varphi_{xx})_{m}^{n} + k/2 (\varphi_{xxt})_{m}^{n} + O(k^2) + O(h^2)
$

</center>

Mirando el termino $R_{hk}f$:

<center>

$
R_{hk}f =\frac{f_{m}^{n+1} + f_{m}^{n}}{2}
$

</center>

Mirando la expación de Taylor tenemos que:

<center>

$
f_{m}^{n+1} = f_{m}^{n} + k (f_{t})_{m}^{n} + k^2/2 (f_{tt})_{m}^{n} + O(k^3)
$


$
f_{m}^{n+1} + f_{m}^{n} = 2 f_{m}^{n} + k (f_{t})_{m}^{n} + k^2/2 (f_{tt})_{m}^{n} + O(k^3)
$

$
\frac{f_{m}^{n+1} + f_{m}^{n}}{2} =  f_{m}^{n} + k/2 (f_{t})_{m}^{n} + O(k^2)
$

</center>

Siendo $f = \varphi_t - b \varphi_{xx}$, tenemos que:

<center>

$
R_{hk}(P\varphi) =  (\varphi_t)_{m}^{n} - b (\varphi_{xx})_{m}^{n} + k/2 ((\varphi_{tt})_{m}^{n} - b (\varphi_{xxt})_{m}^{n}) + O(k^2)
$

$
R_{hk}(P\varphi) =  (\varphi_t)_{m}^{n} - b(\varphi_{xx})_{m}^{n} + \frac{k}{2}(\varphi_{tt})_{m}^{n} - \frac{bk}{2}(\varphi_{xxt})_{m}^{n} + O(k^2)
$

</center>

Mirando los terminos: 

* $ P_{hk}\varphi$:

<center>

$
P_{hk}\varphi = \frac{\varphi_{m}^{n+1} - \varphi_{m}^{n}}{k} - \frac{b}{2h^2} \left(\varphi_{m+1}^{n+1} - 2\varphi_{m}^{n+1} + \varphi_{m-1}^{n+1} \right) - \frac{b}{2h^2} \left(\varphi_{m+1}^{n} - 2\varphi_{m}^{n} + \varphi_{m-1}^{n}\right)

$

$
 = (\varphi_{t})_{m}^{n} + \frac{k}{2} (\varphi_{tt})_{m}^{n} -  \frac{b}{2}  (\varphi_{xx})_{m}^{n} - \frac{bk}{2} (\varphi_{xxt})_{m}^{n} -\frac{b}{2} (\varphi_{xx})_{m}^{n} + O(h^2) + O(k^2)
$

$
 = (\varphi_{t})_{m}^{n} + \frac{k}{2} (\varphi_{tt})_{m}^{n} -  b (\varphi_{xx})_{m}^{n} - \frac{bk}{2} (\varphi_{xxt})_{m}^{n} + O(h^2) + O(k^2)
$

</center>

* $R_{hk}(P\varphi)$:

<center>

$
R_{hk}(P\varphi) =  (\varphi_t)_{m}^{n} - b(\varphi_{xx})_{m}^{n} + \frac{k}{2}(\varphi_{tt})_{m}^{n} - \frac{bk}{2}(\varphi_{xxt})_{m}^{n} + O(k^2)
$

</center>

Tenemos que $ P_{hk}\varphi - R_{hk}(P\varphi)$:

<center>

$
P_{hk}\varphi - R_{hk}(P\varphi) = (\varphi_{t})_{m}^{n} + \frac{k}{2} (\varphi_{tt})_{m}^{n} -  b (\varphi_{xx})_{m}^{n} - \frac{bk}{2} (\varphi_{xxt})_{m}^{n} + O(h^2) + O(k^2) - (\varphi_t)_{m}^{n} + b(\varphi_{xx})_{m}^{n} - \frac{k}{2}(\varphi_{tt})_{m}^{n} + \frac{bk}{2}(\varphi_{xxt})_{m}^{n} + O(k^2)
$

</center>

Nos queda: 

<center>

$
P_{hk}\varphi - R_{hk}(P\varphi) =  O(h^2) + O(k^2)
$

</center>

Por lo tanto el esquema propuesto es de orden $(2,2)$

El esquema de Crank-Nicolson va a estar dado por:

$$
\frac{V_{m}^{n+1} - V_{m}^{n}}{k}
- \frac{1}{2} b \frac{V_{m+1}^{n+1} - 2V_{m}^{n+1} + V_{m-1}^{n+1}}{h^{2}}
- \frac{1}{2} b \frac{V_{m+1}^{n} - 2V_{m}^{n} + V_{m-1}^{n}}{h^{2}}
= \frac{1}{2} \left( f_{m}^{n+1} + f_{m}^{n} \right)
$$

Los valores conocidos son los $n$ y los deconocidos son los $n+1$. Juntando los valores $V_{m}^{n+1}$ y $V_{m}^{n}$:

<center>

$
\left(1 + 2 \frac{bk}{2h^2} \right) V_{m}^{n+1} -  \frac{bk}{2h^2} V_{m+1}^{n+1} - \frac{bk}{2h^2} V_{m-1}^{n+1} -  \left(1 - 2 \frac{bk}{2h^2} \right) V_{m}^{n} - \frac{bk}{2h^2} V_{m+1}^{n} - \frac{bk}{2h^2} V_{m-1}^{n} = \frac{k}{2} \left( f_{m}^{n+1} + f_{m}^{n} \right)
$

</center>

Dejando a un lado los valores desconocidos $(n+1)$ y los valores conocidos $(n)$ se obtiene:

<center>

$
\left(1 + 2 \frac{bk}{2h^2} \right) V_{m}^{n+1} -  \frac{bk}{2h^2} V_{m+1}^{n+1} - \frac{bk}{2h^2} V_{m-1}^{n+1} =  \left(1 - 2 \frac{bk}{2h^2} \right) V_{m}^{n} + \frac{bk}{2h^2} V_{m+1}^{n} + \frac{bk}{2h^2} V_{m-1}^{n} + \frac{k}{2} \left( f_{m}^{n+1} + f_{m}^{n} \right)
$

</center>

Los tiempos y pasos espaciales van a ser $m = 1, ...., M-1$ y $n = 0,...,N-1$:

* Para $n = 0$:

<center>

$
\left(1 + 2 \frac{bk}{2h^2} \right) V_{m}^{1} -  \frac{bk}{2h^2} V_{m+1}^{1} - \frac{bk}{2h^2} V_{m-1}^{1} =  \left(1 - 2 \frac{bk}{2h^2} \right) V_{m}^{0} + \frac{bk}{2h^2} V_{m+1}^{0} + \frac{bk}{2h^2} V_{m-1}^{0} + \frac{k}{2} \left( f_{m}^{1} + f_{m}^{0} \right)

* $m = 1$:

<center>

<center>

\dfrac{b k}{2 h^2}(V_0^{n+1} + V_0^n) = (\frac{bk}{h(2h \alpha -3)}) g_1(t_{n+1}) - \frac{2bk}{h^2(2h \alpha -3)} V_1^{n+1} + \frac{bk}{2h^2(2h \alpha -3)} V_2^{n+1} + (\frac{bk}{h(2h \alpha -3)}) g_1(t_{n}) - \frac{2bk}{h^2(2h \alpha -3)} V_1^n + \frac{bk}{2h^2(2h \alpha -3)} V_2^n

El lado izquierdo de la igualdad y la matriz $A$ matrices quedan:

```python
import numpy as np
import matplotlib.pyplot as plt
from abc import ABC, abstractmethod
```

```python
# Algoritmos para la factorización de matrices y sustitución progresiva y regresiva
# ------ Factorización LU ------
def matrix_factorization_nxn(
    diagonal: np.array,  # Vector que contiene los coeficientes de la matriz (diagonal)
    diagonal_superior: np.array,  # Vector que contiene los coeficientes de la matriz (diagonal superior)
    diagonal_inferior: np.array,  # Vector que contiene los coeficientes de la matriz (diagonal inferior)
):
    """
    Función para realizar la factorización de una matriz tridiagonal A = L*U
    donde L es una matriz triangular inferior con 1's en la diagonal principal y
    U es una matriz triangular superior.
    Parámetros:
    diagonal : np.array
        Vector que contiene los coeficientes de la matriz (diagonal)
    diagonal_superior : np.array
        Vector que contiene los coeficientes de la matriz (diagonal superior)
    diagonal_inferior : np.array
        Vector que contiene los coeficientes de la matriz (diagonal inferior)
    Retorna:
    l_diag : np.array
        Vector que contiene los coeficientes de la matriz L (diagoanl)
    l_diag_inf : np.array
        Vector que contiene los coeficientes de la matriz L (diagonal inferior)
    u_diag_sup : np.array
        Vector que contiene los coeficientes de la matriz U (diagonal)
    """
    l_diag = np.zeros(len(diagonal))
    l_diag_inf = np.zeros(len(diagonal_inferior))
    u_diag_sup = np.zeros(len(diagonal_superior))

    l_diag[0] = diagonal[0]
    l_diag_inf = diagonal_inferior.copy()
    for i in range(0, len(diagonal_superior)):
        u_diag_sup[i] = diagonal_superior[i] / l_diag[i]
        l_diag[i + 1] = diagonal[i + 1] - l_diag_inf[i] * u_diag_sup[i]

    return l_diag, l_diag_inf, u_diag_sup


# ------ Algoritmos para la sustitución progresiva y regresiva ------
# === Sustirución Regresiva ===
# Solución del sistema Ux = z
def sustitucion_regresiva(
    u_diagonal_superior: np.array,  # Vector que contiene los coeficientes de la matriz U (diagonal superior)
    termino_independiente: np.array,  # Vector que contiene los términos independientes
):
    x = np.zeros(len(termino_independiente))
    x[-1] = termino_independiente[-1]
    for i in range(len(termino_independiente) - 2, -1, -1):  # de n-2 hasta 0
        x[i] = termino_independiente[i] - u_diagonal_superior[i] * x[i + 1]

    return x


# === Sustitución Progresiva ===
# Solución del sistema Lz = b
def sustitucion_progresiva(
    l_diagonal: np.array,  # Vector que contiene los coeficientes de la matriz L (diagonal)
    l_diagonal_inferior: np.array,  # Vector que contiene los coeficientes de la matriz L (diagonal inferior)
    termino_independiente: np.array,  # Vector que contiene los términos independientes
):
    z = np.zeros(len(termino_independiente))
    z[0] = termino_independiente[0] / l_diagonal[0]
    for i in range(1, len(termino_independiente)):
        z[i] = (
            termino_independiente[i] - l_diagonal_inferior[i - 1] * z[i - 1]
        ) / l_diagonal[i]

    return z


# Multiplicar la matriz con el vector
def matXvec(diagonal, diagonal_superior, diagonal_inferior, vector):
    """
    Multiplica una matriz tridiagonal A (con diagonales a,b,c) por un vector v.
    Sin construir A.
    """
    res = np.zeros_like(vector)
    for i in range(len(vector)):
        res[i] = diagonal[i] * vector[i]
        if i > 0:
            res[i] += diagonal_inferior[i - 1] * vector[i - 1]
        if i < len(vector) - 1:
            res[i] += diagonal_superior[i] * vector[i + 1]
    return res


class Esquema(ABC):
    def __init__(self):
        pass

    @abstractmethod
    def calcular_esquema(self):
        pass
```

```python
# Constantes del PVI
l = 2  # noqa: E741
T = 1.5
alpha = 2
beta = -3

b = 1

# Definición de las variables de discretización
h = 1 / 100
k = 1 / 1500
```

```python
# Definición de la funciones a utilizar
def solucion_exacta(t, x):
    return np.exp(-t / 2) * np.sin(2 * x - np.pi / 3)


def f(t, x):
    return np.exp(-t / 2) * np.sin(2 * x - np.pi / 3) * (4 * b - 1 / 2)


def u0(x):
    return np.sin(2 * x - np.pi / 3)


def g1(t):
    return np.exp(-t / 2) * (-alpha * np.sin(np.pi / 3) + 2 * np.cos(np.pi / 3))


def g2(t):
    return np.exp(-t / 2) * (
        np.sin(2 * l - np.pi / 3) + 2 * beta * np.cos(2 * l - np.pi / 3)
    )
```

```python
# Mostrar la solución exacta
x_vals = np.arange(0, l + h, h)
t_vals = np.arange(0, T + k, k)

plt.figure(figsize=(10, 6))
plt.plot(x_vals, solucion_exacta(0, x_vals), label="t=0")
plt.plot(x_vals, solucion_exacta(T / 2, x_vals), label="t=T/2")
plt.plot(x_vals, solucion_exacta(T, x_vals), label="t=T")
plt.title("Solución Exacta del PVI")
plt.xlabel("x")
plt.ylabel("u(x,t)")
plt.legend()
plt.grid()
plt.show()
```

```python
# Definición del esquema de diferencias finitas
class EsquemaCrankNicolson(Esquema):
    def __init__(self, valor_h, valor_k, condiciones_iniciales):
        self.valor_h = valor_h
        self.valor_k = valor_k
        self.condiciones_iniciales = condiciones_iniciales

    def calcular_esquema(self):
        # Definición de la discretización de las variables continuas
        x_aux = np.arange(0, l + self.valor_h, self.valor_h)
        t_aux = np.arange(0, T + self.valor_k, self.valor_k)
        # print(f"Tamaño de x: {len(x_aux)} y t: {len(t_aux)}")
        # print("La variable x: ", x_aux)
        # print("La variable t: ", t_aux)

        # Cantidad de puntos
        M = len(x_aux)
        N = len(t_aux) - 1
        # print(f"Tamaño de M: {M} y N: {N}")

        # Matriz tridiagonal
        n_interior = M - 2  # Puntos interiores (excluye x=0 y x=l)
        diag = np.ones(n_interior) + (b * self.valor_k) / (
            self.valor_h**2
        )  # Diagonal de la matriz tridiagonal
        diag_inf = np.zeros(n_interior - 1) - (b * self.valor_k) / (
            2 * self.valor_h**2
        )  # Diagonal inferior de la matriz tridiagonal
        diag_sup = np.zeros(n_interior - 1) - (b * self.valor_k) / (
            2 * self.valor_h**2
        )  # Diagonal superir de la matriz tridiagonal

        # Adición de las condiciones de frontera en la matriz tridiagonal
        diag[0] = diag[0] + (2 * b * self.valor_k) / (
            self.valor_h**2 * (2 * self.valor_h * alpha - 3)
        )
        diag[-1] = diag[-1] - (2 * b * self.valor_k * beta) / (
            self.valor_h**2 * (2 * self.valor_h + 3 * beta)
        )

        diag_sup[0] = diag_sup[0] - (b * self.valor_k) / (
            2 * self.valor_h**2 * (2 * self.valor_h * alpha - 3)
        )
        diag_inf[-1] = diag_inf[-1] + (b * self.valor_k * beta) / (
            2 * self.valor_h**2 * (2 * self.valor_h + 3 * beta)
        )

        # Factorización de la matriz tridiagonal
        diag_L, diag_inf_L, diag_sup_U = matrix_factorization_nxn(
            diagonal=diag, diagonal_superior=diag_sup, diagonal_inferior=diag_inf
        )

        # Valores que multiplican los terminos anteriores
        diag_anterior = np.ones(n_interior) - 2 * (b * self.valor_k) / (
            2 * self.valor_h**2
        )
        diag_inf_anterior = np.zeros(n_interior - 1) + (b * self.valor_k) / (
            2 * self.valor_h**2
        )
        diag_sup_anterior = np.zeros(n_interior - 1) + (b * self.valor_k) / (
            2 * self.valor_h**2
        )

        # Adicion de las condiciones de frontera en la matriz tridiagonal anterior
        diag_anterior[0] = diag_anterior[0] - (2 * b * self.valor_k) / (
            self.valor_h**2 * (2 * self.valor_h * alpha - 3)
        )
        diag_anterior[-1] = diag_anterior[-1] + (2 * b * self.valor_k * beta) / (
            self.valor_h**2 * (2 * self.valor_h + 3 * beta)
        )

        diag_sup_anterior[0] = diag_sup_anterior[0] + (b * self.valor_k) / (
            2 * self.valor_h**2 * (2 * self.valor_h * alpha - 3)
        )
        diag_inf_anterior[-1] = diag_inf_anterior[-1] - (b * self.valor_k * beta) / (
            2 * self.valor_h**2 * (2 * self.valor_h + 3 * beta)
        )

        # Matriz solución
        V = np.zeros((N + 1, M))
        # print("Tamaño de la matriz solución V: ", V.shape)

        # Condiciones iniciales
        V[0, :] = self.condiciones_iniciales(x_aux)

        # Iteración temporal
        for i in range(1, N + 1):
            # Vector F (término fuente)
            F = np.zeros(n_interior)
            for j in range(n_interior):
                f1 = f(t_aux[i - 1], x_aux[j + 1])
                f2 = f(t_aux[i], x_aux[j + 1])  
                F[j] = self.valor_k * (f1 + f2) / 2

            # Condiciones de frontera
            F[0] = F[0] + (
                (b * self.valor_k) / (self.valor_h * (2 * self.valor_h * alpha - 3))
            ) * (g1(t_aux[i]) + g1(t_aux[i - 1]))  # Corrección limite inferior
            F[-1] = F[-1] + (
                (b * self.valor_k)
                / (
                    self.valor_h * (2 * self.valor_h + 3 * beta)
                )  # Corregido denominador
            ) * (g2(t_aux[i]) + g2(t_aux[i - 1]))  # Corregido limite superior

            # Multiplicación de la matriz por los valores en el tiempo n-1
            V_anterior = matXvec(
                diagonal=diag_anterior,
                diagonal_inferior=diag_inf_anterior,
                diagonal_superior=diag_sup_anterior,
                vector=V[i - 1, 1 : M - 1],
            )

            # Vector del lado derecho
            V_anterior += F

            # Resolver el sistema lineal
            z = sustitucion_progresiva(diag_L, diag_inf_L, V_anterior)
            V[i, 1 : M - 1] = sustitucion_regresiva(diag_sup_U, z)

            # Recuperando las condiciones de frontera V_0, V_M
            V[i, 0] = (
                ((2 * self.valor_h) / (2 * self.valor_h * alpha - 3)) * g1(t_aux[i])
                - (4 / (2 * self.valor_h * alpha - 3)) * V[i, 1]
                + (1 / (2 * self.valor_h * alpha - 3)) * V[i, 2]
            )

            V[i, -1] = (
                ((2 * self.valor_h) / (2 * self.valor_h + 3 * beta)) * g2(t_aux[i])
                + (4 * beta / (2 * self.valor_h + 3 * beta)) * V[i, -2]
                - (beta / (2 * self.valor_h + 3 * beta)) * V[i, -3]
            )

        return V
```

```python
esquema = EsquemaCrankNicolson(
    valor_h=h,
    valor_k=k,
    condiciones_iniciales=u0,
).calcular_esquema()

x = np.arange(0, l + h, h)
```

```python
tiempo = np.linspace(0, T, 6)

plt.figure(figsize=(10, 6))

for i in tiempo:
    indice = int(np.round(i / k))

    plt.plot(
        x,
        esquema[indice, :],
        label=f"Solución numérica t={i:.2f}",
        linestyle="--",
        # color="black"
    )
    plt.plot(
        x,
        solucion_exacta(i, x),
        label=f"Solución exacta t={i:.2f}",
        # color="red",
        alpha=0.7,
    )
plt.legend(loc="center left", bbox_to_anchor=(1, 0.5))
plt.tight_layout()
plt.xlabel("$x$")
plt.ylabel("$u(t,x)$")
plt.title(f"Comparación de soluciones con h: {h:.6f} y k: {k:.6f}")
plt.grid()
plt.show()
```

```python
gamma = 0.6

print("=" * 80)
print("SOLUCIÓN NUMÉRICA - ESQUEMA FTCS IMPLÍCITO")
print("=" * 80)
print(f"\nParámetros: T = {T},  gamma = {gamma}")
print("Solución exacta: u(t,x) = exp(- t/2) * sin(2x - pi/3)")
print("\n" + "=" * 80)

h_values_linear = [1 / 10, 1 / 20, 1 / 40, 1 / 80, 1 / 160]
h_values_text = ["1/10", "1/20", "1/40", "1/80", "1/160"]
errors_linear = []


def error(h, u_numeric, u_exact):
    return np.sqrt(h * np.sum(np.abs(u_exact - u_numeric) ** 2))


def compute_order(error1, error2, h1, h2):
    """Compute convergence order"""
    if error1 == 0 or error2 == 0:
        return 0
    return np.log(error1 / error2) / np.log(h1 / h2)


# Table 1: k = O(h) - Linear relationship
print("\n\nTABLA 1: k = gamma*h (relación lineal)")
print("-" * 80)
print(f"{'O(h)':^15} | {'Error':^20} \t\t| {'Orden':^20}")
print("-" * 80)

for i, h in enumerate(h_values_linear):
    k = gamma * h
    x = np.arange(0, l + h, h)
    esquema = EsquemaCrankNicolson(
        valor_h=h,
        valor_k=k,
        condiciones_iniciales=u0,
    )
    error_tn = error(h, esquema.calcular_esquema()[-1, :], solucion_exacta(T, x))
    errors_linear.append(error_tn)

    if i == 0:
        print(f"{h_values_text[i]} \t\t| {error_tn:^20.10e} \t\t| {'-':^20}")
    else:
        order = compute_order(
            errors_linear[i - 1],
            errors_linear[i],
            h_values_linear[i - 1],
            h_values_linear[i],
        )
        print(f"{h_values_text[i]} \t\t| {error_tn:^20.10e} \t\t| {order:^20.6f}")
```

```python
# Table 2: k = O(h^2) - Quadratic relationship
print("\n\nTABLA 2: k = gamma*h² (relación cuadrática)")
print("-" * 80)
print(f"{'O(h²)':^15} | {'Error':^20} \t\t| {'Orden':^20}")
print("-" * 80)

errors_linear = []

for i, h in enumerate(h_values_linear):
    x = np.arange(0, l + h, h)
    k = gamma * h**2
    esquema = EsquemaCrankNicolson(
        valor_h=h,
        valor_k=k,
        condiciones_iniciales=u0,
    )
    error_tn = error(h, esquema.calcular_esquema()[-1, :], solucion_exacta(T, x))
    errors_linear.append(error_tn)

    if i == 0:
        print(f"{h_values_text[i]} \t\t| {error_tn:^20.10f} \t\t| {'-':^20}")
    else:
        order = compute_order(
            errors_linear[i - 1],
            errors_linear[i],
            h_values_linear[i - 1],
            h_values_linear[i],
        )
        print(f"{h_values_text[i]} \t\t| {error_tn:^20.10f} \t\t| {order:^20.6f}")
```

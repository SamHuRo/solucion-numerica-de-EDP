# 1.

Demuestre que el problema de valor inicial: 

$$

\begin{cases}
u_t(t,x) = b\,u_{xx}(t,x), & 0 < x < \ell,\quad t > 0, \\[6pt]
u(0,x) = u_0(x), & 0 \le x \le \ell,
\end{cases}

$$

donde $b \in  \mathbb{R}^+$ y $u_0$ funciónn conocida, esta “bien propuesto”

El PVI está bien puesto si este cumple:  

$ \forall T > 0  \exist C_T > 0 $ talque:

$$
\int_{-\infty}^{\infty} |u(t,x)|^2 dx \leq C_T \int_{-\infty}^{\infty} |u(0,x)|^2 dx, \forall 0 \leq t \leq T
$$


Utilizando la transformada de Fourier en la parte espacial, vemos que:

$$
\hat{u}_t = -b \omega^2 \hat{u} 
$$

esto sale de las propiedades de la transformada de Fourier, la cual cumple:

* $F[u_t(t,x)] = \hat{u}_t(t,\omega)$

* $F[u_{xx}(t,x)] = (i \omega)^2\hat{u}(t,\omega) = -\omega^2\hat{u}(t,\omega)$ , con $F[\frac{d^n}{dx^n} u(t,x)] = (i \omega)^n \hat{u}(t,w)$

Utilizando los valores iniciales la solución a la ecuación queda de la forma:

$$
\hat{u}(t, \omega) = e^{ -b \omega^2 t} \hat{u}_0(\omega)
$$

Utilizando la transformada inversa de Fourier: 

$$
u(t, x) = \frac{1}{\sqrt{2 \pi}} \int_{-\infty}^{\infty} e^{i \omega x} e^{ -b \omega^2 t} \hat{u}_0(\omega) d\omega
$$

Utilizando la transformada inversa de Fourier, obtenesmos:

$$\hat{u}_0(\omega) = \frac{1}{\sqrt{2\pi}} \int_{-\infty}^{\infty} e^{-i\omega y} u_0(y) \, dy$$

Lo cual nos queda: 

$$
u(t, x) = \frac{1}{\sqrt{2 \pi}} \int_{-\infty}^{\infty} e^{i \omega x} e^{ -b \omega^2 t} \left[\frac{1}{\sqrt{2\pi}} \int_{-\infty}^{\infty} e^{-i\omega y} u_0(y) \, dy \right] d\omega
$$

Organizando los terminos nos queda:

$$u(t,x) = \frac{1}{\sqrt{4\pi}} \int_{-\infty}^{\infty} \left( \frac{1}{\sqrt{\pi}} \int_{-\infty}^{\infty} e^{i\omega(x-y)} e^{-b\omega^2t} d\omega \right) u_0(y) \, dy$$

Mirando la integral entre los parentesis, tenemos que: 

$$
 \frac{1}{\sqrt{\pi}} \int_{-\infty}^{\infty} e^{i\omega(x-y)-bt\omega^2} d\omega
$$




Completando cuadrados en la exponencial: 

<center>

$i\omega(x-y)-bt\omega^2 = - bt \left[ \omega^2 - \frac{i (x-y)}{bt}\omega \right]  $ 

</center>

Llamando $\beta = \frac{i (x-y)}{bt}$:

$ - bt \left[ \omega^2 - \frac{i (x-y)}{bt}\omega \right] =  - bt \left[ \omega^2 - \beta \omega \right]  =  - bt \left[ \omega^2 - \beta \omega  + \frac{\beta^2}{4} - \frac{\beta^2}{4} \right] = - bt \left[ \left(\omega - \frac{\beta}{2} \right)^2 - \frac{\beta^2}{4} \right] = - bt \left(\omega - \frac{\beta}{2} \right)^2 + bt \frac{\beta^2}{4}$

Sustituyendo en la integral se obtiene: 

$$
 \frac{1}{\sqrt{\pi}} \int_{-\infty}^{\infty} e^{- bt \left(\omega - \frac{\beta}{2} \right)^2 + bt \frac{\beta^2}{4}} d\omega
$$

Reemplazando $\beta$:

$$
 \frac{1}{\sqrt{\pi}} \int_{-\infty}^{\infty} e^{- bt \left(\omega - \frac{i (x-y)}{2bt} \right)^2 + \frac{i^2 (x-y)^2}{4bt} } d\omega
$$


Reorganizando nos queda: 

$$
 \frac{e^{\frac{- (x-y)^2}{4bt}}}{\sqrt{\pi}} \int_{-\infty}^{\infty} e^{- bt \left(\omega - \frac{i (x-y)}{2bt} \right)^2} d\omega
$$

La integral se conoce como una integral gaussiana la cual cumple con: 

$$
\int_{-\infty}^{\infty} e^{- a x^2} dx = \sqrt{\frac{\pi}{a}}
$$


Realizando la sustitución de $q = \omega - \frac{i (x-y)}{2bt}$; $dq = d\omega$ y vemos que los limites sde integración no van a cambiar, por lo que la integral nos queda: 

$$
 \frac{e^{\frac{- (x-y)^2}{4bt}}}{\sqrt{\pi}} \int_{-\infty}^{\infty} e^{- bt \left(q \right)^2} dq = \frac{e^{\frac{- (x-y)^2}{4bt}}}{\sqrt{\pi}} \sqrt{\frac{\pi}{bt}} = \frac{e^{\frac{- (x-y)^2}{4bt}}}{\sqrt{bt}} 
$$

Dando como resultado: 

$$u(t, x) = \frac{1}{\sqrt{4\pi bt}} \int_{-\infty}^{\infty} e^{-(x-y)^2/4bt} u_0(y) \, dy$$
